$-E+U_F+IR=0$

$I=\frac{E-U_F}{R}$

```python
E = 4.5 # V
U_F = 1.8 # V
R = 100 # ohm

I = (E - U_F) / R
print(f"Pin current: {I} A")

print(f"Pin power:   {U_F * I} W")
```

$R=U/I \Longrightarrow I = U/R$
$$-V_CC+I(R_pu+R)=0\Longrightarrow I(R_pu+R)=V_CC$$
$I(R_pu+R)=V_CC$
$I=\frac{V_CC}{R_pu+R}$



```python
I = VCC / (R_pu + R)

V_PD2 = I * R

print(f"Current at pin: {I}")
print(f"Voltage at external resistor: {V_PD2}")
```
